```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python

df = pd.read_csv('ray_results.csv')

# Convert to a NumPy array
array_data = df.values 
```

```python
test_loss = array_data[:,2].reshape(4,4)
test_accuracy = array_data[:,3].reshape(4,4)
```

```python

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Plot on the first axes (left)
im1 = axes[0].imshow(test_loss, cmap='hot', interpolation='nearest')
fig.colorbar(im1, ax=axes[0]) 
axes[0].set_title('test_loss')
axes[0].set_xlabel('dropouts')
axes[0].set_ylabel('weight_decay')
axes[0].set_xticks([0, 1, 2, 3])
axes[0].set_xticklabels(['0.1', '0.2', '0.3', '0.4'])
axes[0].set_yticks([0, 1, 2, 3])
axes[0].set_yticklabels(['0.0001', '0.001', '0.005', '0.01'])

# Plot on the second axes (right)
im2 = axes[1].imshow(test_accuracy, cmap='hot', interpolation='nearest')
fig.colorbar(im2, ax=axes[1]) 
axes[1].set_xlabel('dropouts')
axes[1].set_ylabel('weight_decay')
axes[1].set_title('test_accuracy')
axes[1].set_xticks([0, 1, 2, 3])
axes[1].set_xticklabels(['0.1', '0.2', '0.3', '0.4'])
axes[1].set_yticks([0, 1, 2, 3])
axes[1].set_yticklabels(['0.0001', '0.001', '0.005', '0.01'])

# Adjust layout to prevent overlap
plt.tight_layout(pad=4)

```
