```python
import torch
import numpy as np
```

### Implement classes for discrete perceptron (DP) and the network

```python
# implement a discrete perceptron
class DP():
  def __init__(self, weight, bias):
    self.w = torch.tensor(weight)
    self.b = torch.tensor(bias)

  def __call__(self, x):
    u = torch.inner(self.w, x) + self.b
    return (u > 0).float()
```

```python
# implement the discrete perceptron network
def network(input):
    n1 = DP([1.0, 1.0], 1.0) # first neuron in the hidden layer
    n2 = DP([1.0, 2.0], -1.0) # second neuron in the hidden layer
    n3 = DP([-1.0, 1.0], 2.0) # third neuron in the hidden layer
    n = DP([1.0, -1.0, 1.0], -1.5) # output neuron

    y1, y2, y3 = n1(input), n2(input), n3(input)  # return tensors of zero dimension
    
    z = torch.cat((y1.reshape(1), y2.reshape(1), y3.reshape(1)), dim=0)
    y = n(z)
    
    return y1, y2, y3, y
```

### Evaluate the network for different inputs

```python
y1, y2, y3, y = network(torch.tensor([0.0, 0.0]))
print('y1:{}, y2:{}, y3:{}, y:{}'.format(y1.numpy(), y2.numpy(), y3.numpy(), y.numpy()))
```

```python
y1, y2, y3, y = network(torch.tensor([2.0, 2.0]))
print('y1:{}, y2:{}, y3:{}, y:{}'.format(y1.numpy(), y2.numpy(), y3.numpy(), y.numpy()))
```

```python
y1, y2, y3, y = network(torch.tensor([-1.0, 1.0]))
print('y1:{}, y2:{}, y3:{}, y:{}'.format(y1.numpy(), y2.numpy(), y3.numpy(), y.numpy()))
```
