# Example 1.1: implementing a neuron

```python
# import torch
import torch

print("torch:", torch.__version__)
```

### Build a neuron
Suppose an input $x$ to a neuron with weights $W$ and bias $b$, Then ouput $y = f(Wx+b)$ where $f$ is the activation function

```python
# define a class for a neuron
class Neuron():
  def __init__(self):  # constructor, self refers to the object itself
    self.w = torch.tensor([2.5, -0.2, 1.0]) # weights
    self.b = torch.tensor(-0.5) # bias

  def __call__(self, x):  # the object can be used as an input in call method
    u = torch.inner(self.w, x) + self.b  # linear sum
    y = 0.8/(1+torch.exp(-1.2*u)) # activation function
    return u, y
```

### Evaluate

```python
# create an object
neuron = Neuron()
```

```python
u, y = neuron(torch.tensor([0.8, 2.0, -0.5]))

print('u: %.3f, y: %.3f'%(u, y))
```

```python
u, y = neuron(torch.tensor([-0.4, 1.5, 1.0]))

print('u: %.3f, y: %.3f'%(u, y))
```
