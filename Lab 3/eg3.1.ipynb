## Input data

```python
# X: 4 samples, 2 features each — the input matrix that gets fed into u = X·w + b
X = np.array([[0.5, -1.66],[-1.0, -0.51],[0.78, -0.65],[0.04, -0.20]])
print('x:{}'.format(X))
```

## Define a class for a perceptron layer

A perceptron layer maps `no_features` inputs to `no_labels` output neurons.
Each output neuron computes a weighted sum of the inputs (plus a bias), then
squashes it with the sigmoid function:

**Pre-activation (logit):**
$$u = Xw + b$$

- $X$ has shape `(batch, no_features)`
- $w$ has shape `(no_features, no_labels)`
- $b$ has shape `(no_labels,)`, broadcast across the batch
- $u$ has shape `(batch, no_labels)` — one logit per output neuron, per sample

**Activation (sigmoid):**
$$y = \sigma(u) = \frac{1}{1 + e^{-u}}$$

Sigmoid squashes each logit independently into $(0, 1)$. Note this is
**not** softmax — the 3 outputs here don't sum to 1, they're 3 independent
"probabilities", one per output neuron.

```python
class PerceptronLayer():
  def __init__(self, no_features, no_labels):
    # w: (no_features, no_labels) — small random init so neurons don't all start identical
    self.w = torch.tensor(np.random.normal(0., 0.1, (no_features, no_labels)), dtype=torch.double)
    # b: (no_labels,) — small positive init, one bias per output neuron
    self.b = torch.tensor(0.1*np.random.rand(no_labels), dtype=torch.double)

  def __call__(self, x):
    # u = X·w + b   (pre-activation / logits, one per output neuron)
    u = torch.matmul(torch.tensor(x), self.w) + self.b
    # y = sigmoid(u) = 1 / (1 + e^-u)   (independent activations per neuron)
    y = torch.sigmoid(u)
    return u, y
```

## Initialize the given layer

```python
model = PerceptronLayer(2, 3)

print('w: {}, b: {}'.format(model.w.numpy(), model.b.numpy()))
```

## Find activations and outputs of the layer

```python
# Single forward pass through the layer: u = X·w + b, y = sigmoid(u)
# (no training here — weights stay at their random init)
u, y = model(X)

print('u:{}'.format(u.numpy()))
print('y:{}'.format(y.numpy()))
```
