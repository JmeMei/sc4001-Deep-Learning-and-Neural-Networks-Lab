# Example 2.2a: GD of a logistic regression neuron (using gradient descent equations)

```python
# import libraries
import torch
import numpy as np
import matplotlib.pylab as plt
```

```python
# set learning parameters
no_epochs = 500
lr = 0.04

SEED = 10
np.random.seed(SEED)
```

```python
# training data
X = np.array([[1.33, 0.72], [-1.55, -0.01], [0.62, -0.72],
    [0.27, 0.11], [0.0, -0.17], [0.43, 1.2], [-0.97, 1.03], [0.23, 0.45]])
Y = np.array([0, 1, 1, 1, 1, 0, 0, 0]).reshape(8,1)

print('X: {}'.format(X))
print('Y: {}'.format(Y))
```

```python
# plot training data

# Y[:,0]==1 -> boolean mask, True for every row whose label is 1 (class A)
# X[mask, 0] -> x1 values of those rows, X[mask, 1] -> x2 values of those rows
# 'bx' = blue x-markers
plt.figure(1)
plt.plot(X[Y[:,0]==1,0], X[Y[:,0]==1,1],'bx', label ='class A')

# same idea but for label 0 (class B), 'ro' = red circle-markers
plt.plot(X[Y[:,0]==0,0], X[Y[:,0]==0,1],'ro', label='class B')

plt.xlabel('$x_1$')
plt.ylabel('$x_2$')
plt.title('training data')
plt.legend()
plt.show()
```

```python
# create a logistic neuron class
# implements the GD (batch) formulas: u = Xw + b*1_P,  f(u) = 1/(1+e^-u)
class Logistic():
  def __init__(self):
    # w: (2x1) weight vector, randomly initialized; b: scalar bias, starts at 0
    self.w = torch.tensor(np.random.rand(2,1))
    self.b = torch.tensor(0.)

  def __call__(self, x):
    # u = Xw + b*1_P   (pre-activation / logit, one value per sample)
    u = torch.matmul(torch.tensor(x), self.w) + self.b
    # f(u) = 1 / (1 + e^-u)   (sigmoid -> probability that class = 1)
    logits = torch.sigmoid(u)
    
    return u, logits
```

```python
# compute entropy and classification error
def loss(targets, logits): # logits = f_u

  # J = -sum_p [ d_p*log(f(u_p)) + (1-d_p)*log(1-f(u_p)) ]   (cross-entropy loss, summed over all P samples)
  entropy = -torch.sum(torch.tensor(targets)*torch.log(logits) + (1-torch.tensor(targets))*torch.log(1-logits))
  # y = 1(f(u) > 0.5); count how many predicted labels differ from the true targets d
  class_err = torch.sum(torch.not_equal(logits > 0.5, torch.tensor(targets)))

  return entropy, class_err
```

```python
# Train the model
# implements: w <- w + alpha*X^T(d - f(u)),  b <- b + alpha*1_P^T(d - f(u))
def train(model, inputs, targets, learning_rate):
  _, f_u = model(inputs)
  
  # grad_u = -(d - f(u)) = f(u) - d   (note the minus sign, so below we SUBTRACT learning_rate*grad)
  grad_u = -(torch.tensor(targets) - f_u)
  # X^T(d - f(u))  ==  -X^T * grad_u
  grad_w = torch.matmul(torch.transpose(torch.tensor(inputs),0,1), grad_u)
  # 1_P^T(d - f(u))  ==  -sum(grad_u)
  grad_b = torch.sum(grad_u)
  
  # w -= lr * grad_w  is the same as  w += lr * X^T(d - f(u))  since grad_w = -X^T(d-f(u))
  model.w -= learning_rate * grad_w
  # b -= lr * grad_b  is the same as  b += lr * 1_P^T(d - f(u))
  model.b -= learning_rate * grad_b
```

```python
# create a logistic neuron object called model
model = Logistic()

print('w: {}, b: {}'.format(model.w.numpy(), model.b.numpy()))
```

```python
entropy, err = [], []

# gradient descent learning begins
for epoch in range(no_epochs):

  u_, f_u_ = model(X)
  entropy_, err_ = loss(Y, f_u_)
  
  train(model, X, Y, learning_rate=lr)
  
  # for first epoch, print values
  if (epoch == 0):
    print('u:{}'.format(u_.detach().numpy()))
    print('f_u:{}'.format(f_u_.detach().numpy()))
    print('y: %d'.format((f_u_ > 0.5).numpy().astype(int)))
    print('entropy:{}'.format(entropy_))
    print('error:{}'.format(err_))
    print('w: {}, b: {}'.format(model.w.numpy(), model.b.numpy()))

  if epoch%100 == 0:
      print('Epoch %2d:  entropy: %2.5f, error: %d'%(epoch, entropy_, err_))

  entropy.append(entropy_), err.append(err_)
```

```python
# print learned weights and error
print('w: %s, b: %s'%(model.w.numpy(), model.b.numpy()))
print('entropy: %2.5f, error: %d'%(entropy[-1], err[-1]))
```

```python
# plot learning curves
plt.figure(2)
plt.plot(range(no_epochs), entropy)
plt.xlabel('epochs')
plt.ylabel('cross-entropy')
plt.show()
```

```python
plt.figure(3)
plt.plot(range(25), np.array(err)[:25])
plt.xlabel('epochs')
plt.ylabel('classification error')
plt.show()
```

```python
# plot the decision boundary
x1 = np.arange(-2, 2, 0.1)
x2 = -(x1*model.w.numpy()[0] + model.b.numpy())/model.w.numpy()[1]

plt.figure(4)
plt.plot(X[Y[:,0]==1,0], X[Y[:,0]==1,1],'bx', label ='class A')
plt.plot(X[Y[:,0]==0,0],X[Y[:,0]==0,1],'ro', label='class B')
plt.plot(x1, x2, '-')
plt.xlabel('$x_1$')
plt.ylabel('$x_2$')
plt.title('decision boundary')
plt.legend()
plt.show()
```
